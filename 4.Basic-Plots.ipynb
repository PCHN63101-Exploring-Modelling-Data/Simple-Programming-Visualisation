# Basic Plots

One of R’s strengths is its ability to produce high-quality plots relatively easily. In this section, we will introduce four common plotting functions: `plot()`, `hist()`, `boxplot()`, and `barplot()`. In the next section, we will see how to customise and improve these plots.

## Scatter plots `plot()`

When you have two numeric variables, a scatter plot is a quick way to explore the relationship between them. For example, we can investigate whether penguins’ bill lengths and flipper lengths are related:


```r
plot(penguins$bill_len, penguins$flipper_len)
```


The `plot()` function requires an `x` and a `y` variable. The first argument is used for the x-axis, so here it is `penguins$bill_len`. The second argument is used for the y-axis, so here it is `penguins$flipper_len`.

The plot suggests a clear positive relationship: as bill length increases, flipper length also tends to increase.

There is a lot we could customise, but for now we can add a title and clear axis labels:

```r
plot(penguins$bill_len,penguins$flipper_len,
    main = "Bill Length vs Flipper Length",
    xlab = "Bill Length (mm)",
    ylab = "Flipper Length (mm)")
```


We have provided three additional arguments:

- `main` sets the plot title.
- `xlab` sets the x-axis label.
- `ylab` sets the y-axis label.


## Histograms: `hist()`

It is often useful to see how a numeric variable is distributed. A histogram can show the approximate range of values and which values occur most often. Let’s look at the distribution of penguin body mass:


```r
hist(penguins$body_mass)
```

The `hist()` function needs only a numeric vector to produce a histogram. Again, there is plenty we could customise. Here, we add an informative title and axis label, and use more bins to show the distribution in greater detail:

```r
hist(penguins$body_mass,
    main = "Histogram of Body Mass",
    xlab = "Body Mass (g)",
    breaks = 14)
```


As before, we added `main` and `xlab` arguments. We also added the `breaks` argument. A break marks the boundary between one bin and the next. The previous plot used seven bins, while this plot uses fourteen.

R will automatically choose a number of bins, but you can adjust `breaks` to show more or less detail.


## Box plots: `boxplot()`

Box plots are also useful for assessing a distribution, particularly when you want to compare a numeric variable across categories.

```r
boxplot(
  body_mass ~ species,
  data = penguins,
  xlab = "Species",
  ylab = "Body mass (g)",
  main = "Penguin body mass by species"
)
```

In the `boxplot()` call, we use the syntax `body_mass ~ species`, as we did with `aggregate()` last week. In other words, we are grouping body mass by species.

Each box plot shows:

- The median, shown by the thick line inside the box.
- The interquartile range (IQR), represented by the box and covering the 25th to 75th percentiles.
- The whiskers, which extend to the most extreme data points within 1.5 times the IQR.
- Potential outliers, shown as individual points beyond the whiskers.

The plot shows at a glance that Gentoo penguins are considerably heavier than the other two species.

## Bar plots: `barplot()`

Bar plots show counts or other summary values for categories. For example, we can count the number of penguins in each species:




```r
barplot(table(penguins$species))
```

The `barplot()` function expects the heights of the bars as its first argument. Here, `table(penguins$species)` creates a table of counts for each species.

When the heights are supplied as a table or named vector, R uses those names as the labels below each bar. Otherwise, you can use the `names.arg` argument to specify the labels yourself.
