# Summary

This week, you learned how to write R code that makes decisions, repeats tasks, creates reusable functions, and communicates results visually. You used conditional statements such as `if`, `else if`, and `else` to run different code depending on whether a condition is `TRUE` or `FALSE`.

You also learned to use `for` loops to repeat the same operation for each item in a vector, list, or set of rows. When looping over vectors, `seq_along()` provides a safer alternative to `1:length()` because it handles empty vectors correctly.

You then created bespoke functions to package reusable code. Functions can accept inputs, provide default argument values, check whether inputs are appropriate, and return one or more results.

Finally, you explored base R plotting. Functions including `plot()`, `hist()`, `boxplot()`, and `barplot()` can be used to investigate relationships, distributions, group differences, and counts. You also learned how to arrange multiple plots with `par()`, set consistent axis scales with `xlim` and `ylim`, and customise appearance using arguments such as `col`, `pch`, `lty`. Additional layers, including reference lines, points, lines, and legends, can be added with `abline()`, `points()`, `lines()`, and `legend()` to make plots clearer and more informative.
