# For Loops in R

A for loop repeats a set of instructions once for each value in a sequence. They are useful when you need to apply the same operation to every item in a vector, list, or set of rows.



## Basic Structure

The general structure of a `for` loop is:

```r
for (variable in sequence) {
  
  # Code to repeat
  
}
```

- `variable` is a temporary name that changes on each repetition.
- `sequence` is the set of values to work through.
- Code inside `{ }` runs once for every value in the sequence.

For example:

```r
for (i in 1:5) {
  print(i)
}
```

Output:

```r
[1] 1
[1] 2
[1] 3
[1] 4
[1] 5
```

Here, `i` takes the values `1`, `2`, `3`, `4`, and `5`.




## Looping Over Values

You can loop directly through a vector of values.

```r
mood_scores <- c(12, 18, 9, 15)

for (score in mood_scores) {
  print(score)
}
```

Output:

```r
[1] 12
[1] 18
[1] 9
[1] 15
```

This processes each participant's mood score in turn.


## Looping Over Positions

Often, you need the **position** of each value rather than the value alone. Use `seq_along()` to loop over vector indices safely.

```r
anxiety_scores <- c(8, 14, 20, 11)

for (i in seq_along(anxiety_scores)) {
  print(anxiety_scores[i])
}
```
```text
[1] 8
[1] 14
[1] 20
[1] 11
```
In this example:

- `i` is the position: `1`, `2`, `3`, then `4`.
- `anxiety_scores[i]` retrieves the score at that position.

You can use this to report participant numbers:

```r
for (i in seq_along(anxiety_scores)) {
  
  message(
    "Participant ", i,
    " has an anxiety score of ", anxiety_scores[i]
  )
  
}
```

Output:

```r
Participant 1 has an anxiety score of 8
Participant 2 has an anxiety score of 14
Participant 3 has an anxiety score of 20
Participant 4 has an anxiety score of 11
```

```{tip}

Printing a message during each iteration of a loop can be a useful way to track progress. For example, if you are importing and pre-processing large raw datasets for each participant, the process could take minutes—or even days. A console message such as `Participant 1 of 30 complete` gives you immediate feedback that the code is running and helps you estimate how long the remaining steps may take.


```


### Why use `seq_along()`?

Avoid this:

```r
for (i in 1:length(anxiety_scores)) {
  print(anxiety_scores[i])
}
```

If `anxiety_scores` is empty, `length(anxiety_scores)` is `0`, and `1:0` produces `c(1, 0)`. This can cause unexpected results.

Instead, use:

```r
for (i in seq_along(anxiety_scores)) {
  print(anxiety_scores[i])
}
```

If the vector is empty, `seq_along(anxiety_scores)` is also empty, so the loop does not run.




## Storing Results From a Loop

A loop does not automatically save its output. Usually, you create an empty object before the loop and fill it during each repetition.

For example, suppose you want to convert raw anxiety scores into percentages.

```r
raw_scores <- c(12, 18, 15, 20)
maximum_score <- 24

percentage_scores <- numeric(length(raw_scores))

for (i in seq_along(raw_scores)) {
  
  percentage_scores[i] <- (raw_scores[i] / maximum_score) * 100
  
}

percentage_scores
```

Output:

```r
[1] 50.00000 75.00000 62.50000 83.33333
```

The line below creates a numeric vector of the correct length:

```r
percentage_scores <- numeric(length(raw_scores))
```

```r
percentage_scores[i] <- ...
```


## For Loops Versus Vectorised R Code

For some of the simple examples above you (e.g., calculating a percentage) you would be much better off using R's vectorised functions rather than a for loop. 

For example, calculate percentages without a loop:

```r
raw_scores <- c(12, 18, 15, 20)
maximum_score <- 24

(raw_scores / maximum_score) * 100
```

Output:

```r
[1] 50.00000 75.00000 62.50000 83.33333
```
You may encounter loops in other people's code, or you may find your self doing some complex analysis for which there doesn't seem to be a vectorised solution. In which case For loops give you the flexibility to create something bespoke.


| Approach | Best use |
|---|---|
| **For loop** | Complex multi-step logic, checks, messages, or calculations that depend on previous iterations |
| **Vectorised code** | Simple calculations applied independently to every value |
| **`apply()` family** | Applying a function across rows or columns |
