# Introduction: Simple Programming and Visualisation

This week, we shift focus slightly to two key topics.

First, we will explore important programming features, including if statements, for loops, and writing custom functions. We will then move on to data visualisation and R’s range of plotting functions. Visualisation is one of R’s greatest strengths and one of the most creative and enjoyable aspects of coding.

By the end of this week, you will have covered all the fundamental concepts of R. It may feel like a lot of information, and you may not yet feel confident applying everything independently. That is completely normal: confidence develops through repeated practice.

We have covered the essentials, but there is still a great deal more to learn. Be prepared to continue developing your skills throughout your time using R.

We have one more week before moving on to statistical modelling. Next week, we will take a slightly different approach by introducing the tidyverse: a collection of packages designed to make data wrangling, summarising, and visualising easier and more efficient. We will also look at R Markdown, which allows you to turn your analysis scripts into clearly presented reports

As you work through this week's content, you can try out the code snippets in your sandbox. Head over to posit cloud here: https://posit.cloud/spaces/797313/content/
